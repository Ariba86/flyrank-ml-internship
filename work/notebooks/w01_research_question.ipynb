```python
import pandas as pd

# Load the dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Show 3 useful numbers
print("Total pages:", len(df))
print("Declining pages:", (df["trend_direction"] == "down").sum())
print("Correlation between search volume and impressions:",
      round(df["search_volume"].corr(df["impressions_90d"]), 3))
```

Quick Look at the Data

The starter dataset contains thousands of webpages for analysis. It includes pages with both growing and declining trends. The correlation between search volume and impressions is very low, showing that search volume alone is not a reliable indicator of page traffic. These observations suggest that multiple factors should be considered when deciding which pages need content updates, making this a suitable machine learning problem.
