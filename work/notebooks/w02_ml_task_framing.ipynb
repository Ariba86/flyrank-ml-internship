## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd

df = pd.read_csv(
    "student-mat.csv",
    sep=";"
)

df.head()
```

```python
metric = "F1-score"

print("Evaluation Metric:", metric)
```

## Unit of Analysis

One row represents one student.

Each record contains information about one student's academic background, study habits, personal factors, and final grade.

The dataframe includes features such as:
- Study time
- Failures
- Absences
- Previous grades
- Family and personal information

```python
def create_performance_level(grade):
    if grade >= 15:
        return "High Performance"
    elif grade >= 10:
        return "Medium Performance"
    else:
        return "Low Performance"


df["performance_level"] = df["G3"].apply(create_performance_level)

df[["G3", "performance_level"]].head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML beats a fixed rule

A simple rule-based system could classify students only using final grades, but student performance depends on multiple factors.

Factors such as study time, failures, absences, family support, and lifestyle can influence academic outcomes.

ML can learn patterns from multiple features and identify relationships that are difficult to capture with a single if-statement.
